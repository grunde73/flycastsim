# Step 1: Simple 1-D model

The simplest model of a cast is the *brick-spring-car* model: a car tows
a brick over a frictionless surface through a linear spring. The car
mimics the rod leverage, the spring the bending and unbending of the rod,
and the brick the fly line.

This step is standalone. It does not use the rod, line or casting
parameters of the later steps.

The `flycastsim` workflow has five steps. Step 1 stands on its own, and
steps 2 to 4 each produce parameters that step 5 uses.

1. Simple 1-D model (`01_simple_1d_model.ipynb`)
2. Rod characteristics (`02_rod_characteristics.ipynb`)
3. Line and leader characteristics (`03_line_and_leader.ipynb`)
4. Casting parameters (`04_casting_parameters.ipynb`)
5. Simulate a fly cast (`05_simulate_cast.ipynb`)

`complete_workflow.ipynb` runs steps 2 to 5 in sequence.

```python
from flycastsim import brick_spring_simple, plot_brick_spring
```

## Run a simulation

The car speed follows a triangular profile: it rises linearly to a peak
and falls back. The simulation stops when the brick overtakes the car.

```python
k = 1.0                # spring stiffness [N/m]
m = 0.01               # brick mass [kg]
d0 = 0                 # initial spring extension [m]
start_cond = [0, 0]    # initial brick position and speed [x(t0), v(t0)]
ts = [1.0, 0.3, 0.45]  # [end of simulation, car speed peak, car stop] [s]
vct = [0, 18, 0]       # car speed at those times [m/s]

# The result is a time-indexed pandas DataFrame.
results = brick_spring_simple(k, m, d0, start_cond, ts, vct)
results.head()
```

## Plot the result

```python
plot_brick_spring(results, plot_cols=results.columns[0:3])
```

## Vary the spring stiffness

A stiffer spring (a stiffer rod) changes when the brick is released and
how fast it is going. The last row of each result is the moment the brick
overtakes the car.

```python
for stiffness in (0.5, 1.0, 2.0):
    res = brick_spring_simple(stiffness, m, d0, start_cond, ts, vct)
    print(f"k = {stiffness:3.1f} N/m: simulation ends at "
          f"t = {res.index[-1]:.3f} s")
```
